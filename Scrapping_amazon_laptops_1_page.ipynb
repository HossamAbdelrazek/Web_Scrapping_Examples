```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.keys import Keys
import time
import numpy as np
```

```python
driver = webdriver.Chrome()
```

```python
driver.get("https://www.amazon.eg/s?k=Gaming+laptop&language=en&_encoding=UTF8")
```

```python
def scroll_down():
    end_height = driver.execute_script("return document.body.scrollHeight")
    
    while True:
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(2)
        new_height = driver.execute_script("return document.body.scrollHeight")
        if new_height == end_height:
            break
        end_height = new_height
```

```python
scroll_down()
Elements = driver.find_elements(By.XPATH, "/html/body/div[1]/div[1]/div[1]/div[1]/div/span[1]/div[1]/div[contains(@role, 'listitem')]")
print(len(Elements))
```

```python
titles = []
image_links = []
prices = []
ratings = []
for element in Elements:
    image_links.append(element.find_element(By.XPATH, ".//img[contains(@class, 's-image')]").get_attribute("src"))
    titles.append(element.find_element(By.XPATH, ".//h2[contains(@aria-label, '')]").get_attribute("aria-label"))
    prices.append(int(''.join(element.find_element(By.XPATH, ".//span[contains(@class, 'a-price-whole')]").text.split(','))))
    try:
        rating_element = element.find_element(By.XPATH, ".//a[contains(@class, 'a-popover-trigger')]").get_attribute("aria-label")
        rating = rating_element.split()[0]
        ratings.append(float(rating))
    except:
        rating = "No rating"
        ratings.append(np.nan)
    
print(len(titles))
print(len(image_links))
print(len(prices))
print(len(ratings))    
    
```

```python
data = {"title": titles, "price": prices, "images": image_links, "rating": ratings}
data
```

```python
import csv
```

```python
csv_file = "amazon_laptops_data.csv"
with open(csv_file, mode="w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(data.keys())
    rows = zip(*data.values())  # Transpose values to match column structure
    writer.writerows(rows)

print(f"CSV file '{csv_file}' created successfully.")
```
