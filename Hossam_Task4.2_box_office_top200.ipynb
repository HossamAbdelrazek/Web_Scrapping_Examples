```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.keys import Keys
import time
```

```python
chrome_options = Options()
# chrome_options.add_argument("--headless")  # Run in headless mode
```

```python
driver = webdriver.Chrome()
```

```python
driver.get("https://www.boxofficemojo.com/chart/top_lifetime_gross/?area=XWW")
```

```python
def scroll_down():
    end_height = driver.execute_script("return document.body.scrollHeight")
    
    while True:
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(2)
        new_height = driver.execute_script("return document.body.scrollHeight")
        if new_height == end_height:
            break
        end_height = new_height
```

Movies Titles Xpath:
/html/body/div[1]/main/div/div/div[5]/div/table[2]/tbody/tr/td[2]/a

LifetimeGross Xpath:
/html/body/div[1]/main/div/div/div[5]/div/table[2]/tbody/tr/td[3]

release_year Xpath: /html/body/div[1]/main/div/div/div[5]/div/table[2]/tbody/tr/td[4]/a

```python
def scrape_page():
    
    scroll_down()
    movie_title = driver.find_elements(By.XPATH, "/html/body/div[1]/main/div/div/div[5]/div/table[2]/tbody/tr/td[2]/a")
    life_time_Gross = driver.find_elements(By.XPATH, "/html/body/div[1]/main/div/div/div[5]/div/table[2]/tbody/tr/td[3]")
    release_year = driver.find_elements(By.XPATH, "/html/body/div[1]/main/div/div/div[5]/div/table[2]/tbody/tr/td[4]/a")
    titles = []
    gros = []
    years = []
    for i in range(len(movie_title)):
        titles.append(movie_title[i].text)
        gros.append(life_time_Gross[i].text)
        years.append(release_year[i].text)
    return titles, gros, years
```

```python
titles, gros, years = scrape_page()
data = {'title': titles, 'lifetime_gross': gros, 'release_date': years}
```

```python
import csv
```

```python
csv_file = "Box_office_top200.csv"
with open(csv_file, mode="w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(data.keys())
    rows = zip(*data.values())  # Transpose values to match column structure
    writer.writerows(rows)

print(f"CSV file '{csv_file}' created successfully.")
```
