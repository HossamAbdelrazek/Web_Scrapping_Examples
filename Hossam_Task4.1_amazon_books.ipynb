```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.keys import Keys
import time
```

```python
chrome_options = Options()
# chrome_options.add_argument("--headless")  # Run in headless mode
```

```python
driver = webdriver.Chrome()
```

```python
driver.get("https://www.amazon.in/gp/bestsellers/books/")
```

```python
def scroll_down():
    end_height = driver.execute_script("return document.body.scrollHeight")
    
    while True:
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(2)
        new_height = driver.execute_script("return document.body.scrollHeight")
        if new_height == end_height:
            break
        end_height = new_height
```

```python
def scrape_page():
    
    scroll_down()
    Titles = driver.find_elements(By.XPATH, "/html/body/div[1]/div[2]/div/div/div[1]/div/div/div[2]/div[1]/div[1]/div/div/div/div[2]/span/div/div/div/a/span/div")
    prices = driver.find_elements(By.XPATH, "/html/body/div[1]/div[2]/div/div/div[1]/div/div/div[2]/div[1]/div[1]/div/div/div/div[2]/span/div/div/div/div[4]/div/div/a/div/span/span")
    page_info = {}
    for title, price in zip(Titles, prices):
        page_info[title.text] = price.text
    return page_info
```

```python
pages = []
while True:
    pages.append(scrape_page())
    
    try:
        next_button = driver.find_element(By.XPATH, '/html/body/div[1]/div[2]/div/div/div[1]/div/div/div[2]/div[2]/ul/li[4]/a[contains(text(), "Next page")]')  # Update XPath
        driver.execute_script("arguments[0].click();", next_button)  # Scroll into view and click
        time.sleep(3)  # Allow time for next page to load
    except:
        print("No more pages.")
        break  # Stop when there's no next button
    
```

```python
titles = []
prices = []
for page in pages:
    for key, val in page.items():
        titles.append(key)
        prices.append(val)
        
data = {"title": titles, "price": prices}
```

```python
import csv
```

```python
csv_file = "amazon_india_books.csv"
with open(csv_file, mode="w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(data.keys())
    rows = zip(*data.values())  # Transpose values to match column structure
    writer.writerows(rows)

print(f"CSV file '{csv_file}' created successfully.")
```
